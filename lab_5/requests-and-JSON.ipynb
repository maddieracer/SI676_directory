3. Using the same three items as above, write a block of Python code using the requests library that will request and retrieve the JSON data about the items above, then save them in local JSON files.

```python
import json
import requests
from pathlib import Path

OUT = Path('..', 'output')

baseURL = 'https://www.loc.gov/'
itemEndpoints = [
    '/resource/cph.3f05183/',
    '/resource/fsa.8d24709/',
    '/resource/highsm.64003/'
]
searchEndpoint = '/search/'

parameters = {
    'fo' : 'json'
}

for itemEndpoint in itemEndpoints:
    r = requests.get(
        baseURL + itemEndpoint,
        params=parameters
    )

print('response code:', r.status_code)
print('response headers:', r.headers)

data = r.json()

r.json()['item']['title']
title = r.json()['item']['title'].split('/')[0]
print(title)

newfile = OUT / f'{title}.json'

OUT.mkdir

with newfile.open('w', encoding='utf-8') as file:
    json.dump(data, file)

print('Tried to write the file!')
```

4. Create a short block of Python code that queries the base URL. Use the input() function so the code will ask the user for a unique query, save that query as a string that will be inserted into the URI request as a parameter, send the request using python requests, and return the results.

```python
import json
import requests

base_url = "https://www.loc.gov/search/"
query = input("Enter a Library of Congress search query: ")


parameters = {
        "q": query,
        "fo": "json"
    }

response = requests.get(
        base_url,
        params=parameters,
    )

response.raise_for_status()

results = response.json()

print(f"\nRequested URI: {response.url}\n")
```
